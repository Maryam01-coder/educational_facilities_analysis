**DATA CLEANING**

```python
#Handling missing values 
df = df.dropna()
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
df.info()
```

**Reason for dropping null values**
I decided to drop all null values in dataset because;
* A very relevant amount of data still remains after the removal as only **12.57%** of the data was removed.
* Filling the null values in some columns with either the mode value or average will affect the analysis of the data and can lead to wrong insights.

```python
#fixing datatypes
df["improved_water_supply"] = df["improved_water_supply"].astype(bool)
df["improved_sanitation"] = df["improved_sanitation"].astype(bool)
df["chalkboard_each_classroom_yn"] = df["chalkboard_each_classroom_yn"].astype(bool)
df["phcn_electricity"] = df["phcn_electricity"].astype(bool)
```

```python
#cleaning date value
from dateutil import parser

def clean_date(date_str):
    try:
        return parser.parse(date_str, dayfirst=True)
    except Exception:
        return pd.NaT

df["date_of_survey"] = df["date_of_survey"].apply(clean_date)
df.info()
```

```python
#droping unnecessary columns
df = df.drop(columns=["formhub_photo_id", "survey_id", "gps", "facility_id"])
df.head()
```

```python
#Validate Numeric Columns
numeric_cols = [
    'num_tchrs_with_nce',
    'num_tchr_full_time',
    'num_students_total',
    'num_classrms_total',
    'num_toilets_total'
]

df[numeric_cols].describe()
```

```python
#checking for negative values 
(df[numeric_cols] < 0).sum()
```

```python
df["ward"].unique()
```

```python
#handling input that are only numbers
df["ward"] = df["ward"].str.replace(
    r"^\s*(\d+)\s*$",
    r"Ward \1",
    regex=True
)
df.head()
```

```python
#handling values with roman numerals
roman_to_number = {
    "I": "1",
    "II": "2",
    "III": "3",
    "IV": "4",
    "V": "5",
    "VI": "6",
    "VII": "7",
    "VIII": "8",
    "IX": "9",
    "X": "10"
}

for roman, number in roman_to_number.items():
    df["ward"] = df["ward"].str.replace(
        rf"^Ward\s+{roman}$",
        f"Ward {number}",
        regex=True
    )
```

```python
#standardizing texts
df["ward"] = (df["ward"]
    .str.strip()
    .str.title()
    .str.replace("Gra", "GRA", regex=False)
)

df["community"] = (df["community"]
    .str.strip()
    .str.title()
    .str.replace("Gra", "GRA", regex=False)
)
```

```python
#creating new columns 
df["student_to_teacher_ratio"] = (df['num_students_total'] / df['num_tchr_full_time']).round(2)
df["total_teachers"] = (df["num_tchrs_with_nce"] + df["num_tchr_full_time"])
df["facility_density"] = df.groupby("ward")["ward"].transform("count")
```

```python
df.head()
```

```python
#creting seperate colums for state and local gorvenment
df[["state", "lga"]] = df["unique_lga"].str.split("_", n=1, expand=True)
df.head()
```

```python
df["state"] = (df["state"]
    .str.strip()
    .str.title()
    .str.replace("Cross", "Cross River", regex=False)
    .str.replace("Rivers", "Cross River", regex=False)
)
df.head()
```

```python
df.to_csv("cleaned_facilities_records.csv", index=False)
```

**Summary Report and Recommendations**

**Key Findings**

The analysis shows that Nigeria has a significant educational infrastructure problem. Only **38.09%** of schools have **improved water supply**, **46.17%** have **improved sanitation**, and just **24.49%** have **electricity**. This means that many schools lack the basic facilities needed to provide a safe and effective learning environment, with electricity being the most serious gap.

Primary education represents the largest share of the facilities analysed, with **25,210 Primary Only schools** and **16,717 Pre-primary and Primary schools**. This suggests that primary education should receive considerable attention when planning infrastructure improvements.

The average school has about 338 students and 18 teachers, but there are large differences between schools, with some having no recorded students and others having more than 2,300 students. This shows that national averages alone can hide problems such as **overcrowding and uneven teacher distribution**.
Public schools carry most of the education burden, serving about **24.82 million students**, compared with **3.96 million** in private schools and about **354,000** in faith-based schools. Public schools also have the largest number of teachers, with approximately 1.19 million, compared with 366,000 in private schools.

However, public schools have significantly poorer infrastructure. Only 32.24% have improved water, 38.33% have improved sanitation, 43.30% have a chalkboard in each classroom, and 16.65% have electricity. In comparison, private schools have 58.23% water, 73.65% sanitation and 52.36% electricity, while faith-based schools have 68.55%, 79.78% and 52.19%, respectively. This shows a clear infrastructure gap between public and non-public schools.
Geographic Findings

Infrastructure also varies greatly across the 772 LGAs in the cleaned dataset. Some LGAs have very high coverage, while others have extremely low levels. For sanitation, Kano Municipal and Sokoto South recorded 100%, while Miga, Jigawa and Ngala, Borno recorded 0%. Several LGAs in Taraba also recorded 0% electricity access.

**What Major Problems Exist?**

The major problems identified are poor electricity access, inadequate water supply, poor sanitation, unequal infrastructure between public and private schools, differences between locations, uneven student-teacher distribution, and limited classroom resources. These problems affect the quality and safety of the learning environment and are particularly serious in underserved communities.

**Which Regions Need the Most Attention?**

The greatest attention should go to underserved LGAs, particularly in parts of Taraba, Borno, Yobe, Jigawa and Plateau, where the analysis found very low or zero access to key facilities. However, needs should be assessed at the LGA level rather than assuming that every school in a state has the same conditions. Lagos and some other LGAs perform better on certain indicators, but this does not mean they have no educational challenges.
Priority should be based on two factors: the severity of the infrastructure gap and the number of students/facilities affected.

**How Governments and NGOs Can Use the Data?**

Governments and NGOs can use the data to make funding and intervention decisions more targeted. Instead of distributing resources equally, they can identify schools and LGAs with the greatest infrastructure gaps and prioritise them.

For example, NGOs can target schools without improved water and sanitation, energy organisations can focus on schools without electricity, education organisations can target schools with large student populations, and teacher-support programmes can focus on schools with high student-to-teacher pressure.

Because public schools serve the majority of students while having poorer infrastructure, they should receive significant attention. Funding can focus on water systems, sanitation, classrooms, teaching materials, electricity/solar power and basic digital infrastructure. Combining infrastructure conditions with student population would also help identify where investments can have the greatest impact.

**Recommendations**
1.	Prioritise water, sanitation and electricity as immediate infrastructure needs.
2.	Give greater attention to public schools, which serve the largest student population but have weaker infrastructure.
3.	Target interventions at the LGA level based on actual need rather than using a one-size-fits-all approach.
4.	Use solar and other renewable energy solutions in areas with poor or unreliable grid electricity.
5.	Combine infrastructure gaps with student population when allocating resources.
6.	Improve teacher planning by considering enrolment and student-teacher pressure.
7.	Develop a school infrastructure priority index to rank schools according to their level of need.
8.	Regularly update the data so that decisions are based on current conditions.
  
**Conclusion**

A major limitation is that the dataset contains survey information from **2010–2014**, so it should not be treated as a complete picture of Nigeria's educational infrastructure today. However, it provides useful evidence of historical infrastructure gaps and demonstrates how data can be used for targeted planning.
Overall, the analysis shows that Nigeria's education system faces serious challenges in electricity, water, sanitation and basic classroom resources, with public schools and several underserved LGAs facing the greatest difficulties. The key takeaway is that education funding should be data-driven and targeted, using school-level and LGA-level information to direct resources to the areas where they are most needed and where they can benefit the largest number of students.
